```python
import torch

tensor = torch.tensor([
    [2, 3, 4, 5],
    [6, 7, 8, 9]
])

ew = tensor * tensor

# shape  - number of rows and columns
rows = tensor.shape[0]
cols = tensor.shape[1]

print("Element-wise multiplication details:\n")

for i in range(rows):
    for j in range(cols):
        a = tensor[i, j]      # current element 
        b = tensor[i, j]      # same element again 
        c = ew[i, j]          # result of multiplication 
        print(f"tensor[{i},{j}] * tensor[{i},{j}] = {a} * {b} = {c}")
```

```python
rows = tensor.shape[0]
cols = tensor.shape[1]
print(f"\nTensor shape: {tensor.shape} (rows: {rows}, columns: {cols})")
```

## 5. In-place operations with underscore _ (اعمال درجا با زیرخط)

In-place operation means that the original variable itself is changed (no new tensor).

```python
tensor = torch.tensor([1, 0, 1, 1, 1, 0, 1, 1])
print("Original tensor :", tensor)

tensor.add_(5)
```

 1) Add 5 to each element of the tensor.
 2) Store the result in the same tensor (in-place operation).

 before : [1, 0, 1, 1, 1, 0, 1, 1]
 
 after:  [6, 5, 6, 6, 6, 5, 6, 6]

```python
print("Tensor after in-place addition:", tensor)
```

## 6. Tensor <-> NumPy and Shared Memory

### 6.1 Tensor -> NumPy

For a CPU tensor, `tensor.numpy()` usually returns a NumPy array that shares the same underlying memory.

Note:
- This section assumes CPU tensors.
- For GPU tensors, move to CPU first (for example, `tensor.cpu().numpy()`).

```python
# Create a 1D tensor of ones

t = torch.ones(5)

# t – tensor with shape (5,) and values ​​[1., 1., 1., 1., 1.]

# Convert to NumPy array

n = t.numpy()

# n – NumPy array.
# Note: t and n share memory.

print("Tensor t:", t)
print("NumPy n:", n)
```

### 6.2 In-place Update on Tensor Affects NumPy (when memory is shared)

If `n = t.numpy()` shares memory with `t`, then an in-place update on `t` (such as `t.add_(1)`) is visible in `n` as well.

```python
t.add_(1) # add 1 in-place
print("Tensor t after add_:", t)
print("NumPy n after tensor change:", n)
# Since t and n share memory, the change in t is also visible to n.
```

### 6.3 NumPy -> Tensor and In-place Update on NumPy

`torch.from_numpy(n2)` creates a tensor that shares memory with `n2`.

Therefore, in-place updates on `n2` are reflected in `t2`, and vice versa.

```python
import numpy as np

# Create NumPy array
n2 = np.ones(5)
print("Original NumPy n2:", n2)

# From NumPy to tensor
t2 = torch.from_numpy(n2)
# t2 – is a tensor created from n2 and has shared memory.

print("Tensor t2:", t2)

# In-place add on NumPy array
np.add(n2, 1, out=n2)
# np.add(a, b, out=a) – adds b to each element of a and writes the result to out (a itself).
# Here: all elements of n2 are incremented by one.

print("NumPy n2 after np.add:", n2)
print("Tensor t2 after np.add on n2:", t2)
# Since t2 and n2 share memory, t2 has also changed.
```
