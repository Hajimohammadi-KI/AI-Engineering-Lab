# Environment Setup Guide

> **Run this guide once on any new machine before opening the notebook.**  
> After setup, select the `.venv-bachelor-ml` kernel in VS Code and run the cells normally.

---

## Step 1 — Create the virtual environment

Open **CMD or PowerShell** inside the project root folder and run:

```bash
python -m venv .venv-bachelor-ml
```

---

## Step 2 — Enable Windows Long Paths (required for PyTorch)

PyTorch contains deeply nested files that exceed Windows' default 260-character path limit.  
**You must enable long paths before installing — otherwise pip will fail.**

Open **PowerShell as Administrator** and run:

```powershell
Set-ItemProperty -Path "HKLM:\SYSTEM\CurrentControlSet\Control\FileSystem" -Name "LongPathsEnabled" -Value 1
```

Then **restart your PC** (or at minimum restart PowerShell/CMD).

> ✅ This is a one-time system setting. You only need to do it once per machine.

---

## Step 3 — Install PyTorch (with GPU support)

Check your CUDA version first:
```bash
nvidia-smi
```

Then install PyTorch using the venv's pip:

```bash
# CUDA 12.4 — works with any CUDA 12.x driver (recommended)
.venv-bachelor-ml\Scripts\pip.exe install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu124

# CPU only (no GPU)
.venv-bachelor-ml\Scripts\pip.exe install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu
```

> **RTX 50xx / Blackwell GPU?** Use the nightly build instead:
> ```bash
> .venv-bachelor-ml\Scripts\pip.exe install --pre torch torchvision torchaudio --index-url https://download.pytorch.org/whl/nightly/cu128
> ```

---

## Step 4 — Install all other packages

```bash
.venv-bachelor-ml\Scripts\pip.exe install grad-cam scikit-learn seaborn matplotlib numpy ipykernel opencv-python tqdm scipy pandas
```

---

## Step 5 — Select the kernel in VS Code

1. Open VS Code in the project folder  
2. Press `Ctrl+Shift+P` → **Python: Select Interpreter**  
3. Choose `.venv-bachelor-ml\Scripts\python.exe`  
4. Open the notebook → top-right kernel picker → select `.venv-bachelor-ml (Python 3.13.x)`
